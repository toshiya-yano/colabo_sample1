●if文  
分岐にはif文を用います。  
ifの条件が満たされていなければ、elifの条件が上から順番に判定されます。  
これらの条件がすべて満たされていなければ、elifの処理が実行されます。  
多くのプログラミング言語では分岐や関数のブロックを表すために()を  
使用しますが、Pythonではブロックの範囲をインデントで表します。  
すなわちインデントしていない行が出現したらその直前にブロックは
終了していることになります。  
インデントには半角スペース4つを用いることが多いです。


```python
a=7
if a < 12:
  print("Good Morning!")
elif a < 17:
  print("Good Afternoon!")
elif a < 21:
  print("Good Evening!")
else:
  print("Good Night!")
```

●for文　  
指定した回数をループするためにはfor文を用います。  
Loopする範囲を指定するためにはListやrangeをin演算子とともに用います。  
rangeの使い方は以下の通りです。
range([開始番号,]終了番号[,ステップ数]  
例えばrange(3)は0から2までの範囲になります。  


```python
for a in [4,7,10]:  #Listを作成⇒List「a」の中に4，7，10が入る
    print(a)
for b in range(3):  #range(3)は0から2までの範囲
    print(b)  #つまり0と1と2が順番に格納されていることを理解してほしい！
```

●Whoile文
ある条件を満たしている間、ループを繰り返す。


```python
a=0
#aが5より小さい間はループを繰り返す。
while a <5:
    print(a)  #まずは初期値0を表示して、下のa+1で1づつ加えて4まで繰り返す？
    a+=1  #5まで行くけど、5の場合はループが終了しているので5が表示されることもないし、それ以降で1が足されることもない！
```

●内包表記  
内包表記はListの要素を操作した上で、新しいListを作成するための記法です。  
通常、そのような処理はforやwhileによるループを使用しますが、  
内包表記を用いると、簡潔に記述することができます。  
内包表記は、以下の形式で記述します。  
新たなList＝[要素への処理 for 要素 in List]
リスト内の要素をひとつ取り出して、要素への処理を実行した上で新しいListを作成します。  


```python
a=[1,2,3,4,5,6]
    #aの要素を3倍して1を足して新たなListを作る！
b=[c*3+1 for c in a]
print(b)
```

●関数  
関数を用いると複数行の処理をまとめることができます。  
関数はdefの後に関数名を記述し、()の中に引数を記述します。  
returnのあとの値が返り値になります。  


```python
#関数の定義
def add(a,b):  #関数の定義：defに続けて関数名を記述し、()内に引数としての変数を記述してやry
  c=a+b
  return c


print(add(3,4))
```

Pythonの引数にはDefault値を設定できます。  
Default値を設定すると、関数を呼び出す際にその引数を省略できます。  
以下の例では、第2引数にDefault値を設定しています。



```python
def add(a,b=4): #default値を設定しています
  c=a+b
  #return c
  print("add=",c)
add(3)  #第2引数は設定しない""

#また*アスタリスクを付けたタプルを用いて、複数の引数を一度に渡すことができる！
def add2(a,b,c):
    d=a+b+c
    print("add2=",d)
e=(1,2,3)
add2(*e)
f=[4,5,6]
add2(*f)

def add3(a,b,c):
  d=a+b+c
  return(d)
g=(7,8,9)
# add3(*g) # この行は単独では何も表示しません
print("add3=",add3(*g))
```

```python


def add3(a,b,c):
  d=a+b+c
  return(d)
g=(7,8,9)
h=add3(*g)
i=h+sum(g) # `g`ではなく、`h`にsum(g)を足すように修正
print("add3:i=",i)
```
