●浮動小数点型の値は死す表記が可能です。  
以下のようにeを用いて小数点以下の値も表記することができます。

```python
print(1.2E5)

num = 1.2e-5
print(num)
# f文字列を使用して小数点以下6桁で表示する
formatted_num = f"{num:.6f}"
print(formatted_num)
# 出力結果: 1.2e-05


number = 1.2e-5
# f文字列を使用して小数点以下6桁で表示する
formatted_number = f"{number:.6f}"

print(formatted_number)
# 出力結果: 0.000012
1.2E5
```

●List　PythonのListは複数の値をまとめて格納することができる。  
Listは全体を[]で囲み、各要素は,で区切ります。  
PythoのListはどのような型でも格納することができ、Listの中にListを格納することができます。  
Listの各要素へのAccessはIndexを使用し、要素の追加や入れ替えなどが可能です。  
要素の対ははappendを使用する  


```python
a=[1,2,3,4,5]
b=a[2]
print("b=",b)
a.append(6)  #末尾に6を追加する
print("a=",a)
#要素の入れ替え⇒2番目の要素を11に入れ替える
a[1]=11 #配列の要素は0番目から始まる！
print("a=",a)
```

●タプル
タプルはListと同じく複数の値をまとめて扱いたいときに利用します。  
要素の追加や変更がない場合は、Listよりもタプルを使用する方がベターです。



```python
#タプルの作成
a=(1,2,3,4,5)
#このタプルのINDEXが2つまり3番目の値を表示するには
b=a[2]
print("b=",b)
```

●辞書  
辞書はKeyと値の組み合わせでDataを格納します。
以下は、Pythonの辞書を扱う例です。  
文字列をKeyとして辞書を作成し、値の取得や入れ替え要素の追加を行ってゆきます。  


```python
#辞書の作成
a={"Artificial":1}  #辞書の作成
print(a["Artificial"]) # キーを修正

b={"artificial":1,"Interigence":2}
print(b["Interigence"])

#要素の入れ替え
b["Interigence"]=7
print(b["Interigence"])

#elementの追加
b["ML"]=3  #要素の追加
print(b)
```

●セット  
セットはListと似ていますが、重複した値を持つことはできない。  
タプルと異なり、要素の削除や追加が可能である。  


```python
a=[1,1,2,3,4,4,5,5,15]
print(a)
b=set(a) #List:aをsetに変換
print(b)
#値を削除
b.remove(15)
b.remove(5)
print(b)
```

●if文  
分岐にはif文を用います。  
ifの条件が満たされていなければ、elifの条件が上から順番に判定されます。  
これらの条件がすべて満たされていなければ、elifの処理が実行されます。  
多くのプログラミング言語では分岐や関数のブロックを表すために()を  
使用しますが、Pythonではブロックの範囲をインデントで表します。  
すなわちインデントしていない行が出現したらその直前にブロックは
終了していることになります。  
インデントには半角スペース4つを用いることが多いです。


```python
a=7
if a < 12:
  print("Good Morning!")
elif a < 17:
  print("Good Afternoon!")
elif a < 21:
  print("Good Evening!")
else:
  print("Good Night!")
```
